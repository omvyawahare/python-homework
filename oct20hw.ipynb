```python
class Student:
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks

s1 = Student(1, "Jay", 68)
s2 = Student(10, "Pavan", 88)
s3 = Student(6, "Viru", 98)
s4 = Student(4, "Priya", 58)
s5 = Student(11, "Payal", 78)

student_list = []

student_list.append(s1)
student_list.append(s2)
student_list.append(s3)
student_list.append(s4)
student_list.append(s5)

for obj in student_list:
    print(obj.name, obj.marks)

```

```python
class Car:
    def __init__(self,brand,colour,price):
        self.brand = brand
        self.colour = colour
        self. price = price
        
c1 = Car("toyota","black",5200000)
c2 = Car("kia","white",1900000)
c3 = Car("bmw","green",8600000)
c4 = Car("tata","black",1500000)
c5 = Car("mahindra","red",1300000)

car_list =[]

car_list.append(c1)
car_list.append(c2)
car_list.append(c3)
car_list.append(c4)
car_list.append(c5)

for obj in car_list:
    print(obj.colour, obj.price)
```

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

e1 = Employee(101, "Amit", 25000)
e2 = Employee(102, "Rahul", 30000)
e3 = Employee(103, "Priya", 35000)
e4 = Employee(104, "Sneha", 40000)
e5 = Employee(105, "Vikas", 45000)

employee_list = []

employee_list.append(e1)
employee_list.append(e2)
employee_list.append(e3)
employee_list.append(e4)
employee_list.append(e5)

for obj in employee_list:
    print(obj.emp_id, obj.name, obj.salary)
```

```python
class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

m1 = Mobile("Apple", "iPhone 17", 80000)
m2 = Mobile("Samsung", "S25", 75000)
m3 = Mobile("OnePlus", "13R", 45000)
m4 = Mobile("Vivo", "V50", 35000)
m5 = Mobile("Oppo", "Reno 13", 30000)

mobile_list = []

mobile_list.append(m1)
mobile_list.append(m2)
mobile_list.append(m3)
mobile_list.append(m4)
mobile_list.append(m5)

for obj in mobile_list:
    print(obj.brand, obj.model, obj.price)
```

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

b1 = BankAccount("Amit", 1001, 25000)
b2 = BankAccount("Rahul", 1002, 35000)
b3 = BankAccount("Priya", 1003, 50000)
b4 = BankAccount("Sneha", 1004, 15000)
b5 = BankAccount("Vikas", 1005, 40000)

account_list = []

account_list.append(b1)
account_list.append(b2)
account_list.append(b3)
account_list.append(b4)
account_list.append(b5)

for obj in account_list:
    print(obj.account_holder, obj.account_number, obj.balance)
```
