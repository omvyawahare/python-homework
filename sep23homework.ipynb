### s = "jay patil"

```python
s = "Jay Patil"

print(s[0:1])    # J
print(s[0:2])    # Ja
print(s[0:3])    # Jay
print(s[0:4])    # Jay 
print(s[0:5])    # Jay P
print(s[0:6])    # Jay Pa
print(s[0:7])    # Jay Pat
print(s[0:8])    # Jay Pati
print(s[0:9])    # Jay Patil

print(s[1:4])    # ay 
print(s[1:5])    # ay P
print(s[1:6])    # ay Pa
print(s[2:5])    # y P
print(s[2:7])    # y Pat
print(s[3:8])    #  Pat
print(s[4:9])    # Patil

print(s[:3])     # Jay
print(s[4:])     # Patil
print(s[-5:])    # Patil
print(s[::-1])   # litaP yaJ
```
