```python

username = input("Enter username: ")
password = input("Enter password: ")
email = input("Enter email: ")

# Validate username
if len(username) > 3:

    # Validate password
    if len(password) >= 8 and any(ch.isdigit() for ch in password):

        # Validate email
        if "@" in email:
            print("Your entered data is correct")

        else:
            print("Invalid email: @ is required")

    else:
        print("Invalid password: Minimum 8 characters and at least one number required")

else:
    print("Invalid username: Must contain more than 3 characters")
```
